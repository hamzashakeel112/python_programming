```python
import matplotlib.pyplot as plt 
```

```python
# chart 
x=[1,2,3,4,5]
y=[110,90,130,140,150]

plt.plot(x,y) # this is line chart 
plt.show()
```

```python
# customize a chart 
plt.figure(figsize=(4,2))
plt.plot(x,y,color="green",marker='o', linestyle="dashed",linewidth=2, markersize=10)
plt.title("Sales chart")
plt.xlabel("months")
plt.ylabel("sales")
```

```python
# multiple charts
x=[1,2,3,4,5]
y1=[20,30,40,50,60]
y2=[25,35,45,55,60]
plt.plot(x,y1,label="sales 2024",color="blue")
plt.plot(x,y2,color="green",label="sales 2025")
plt.title("yoy")
plt.xlabel("month")
plt.ylabel("sales")
plt.legend()
```

```python
# bar chart
plt.figure(figsize=(4,3))
x=[1,2,3,4,5]
y1=[30,40,10,20,50]
# plt.bar(x,y1)
plt.scatter(x,y1)
plt.show()
```
